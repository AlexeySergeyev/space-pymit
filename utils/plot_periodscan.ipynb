```python
import matplotlib.pyplot as plt
import plotly.express as px
import numpy as np
import pandas as pd
import os
```

```python
path = "../gui_output/A153814/period_scan/"
files = os.listdir(path)
files = [f for f in files if f.endswith("txt") and f.startswith("period_scan")]
files.sort()
files
```

```python
df = pd.read_csv(os.path.join(path, files[0]), sep="\s+", header=None, names=["period", "rms", "chi2", "iter", "dark_area"])
df
```

```python
fig = px.line()
for f in files[:]:
    df = pd.read_csv(os.path.join(path, f), sep=r"\s+", header=None, names=["period", "rms", "chi2", "iter", "dark_area"])
    fig.add_scatter(x=df["period"], y=df["chi2"], mode="markers", name=f)
fig.update_layout(title="Period Scan", xaxis_title="Period", yaxis_title="Chi2")
fig.show()
# plt.xlabel("Period")
# plt.ylabel("Chi2")
# plt.title("Period Scan")
# plt.grid(alpha=0.3)
# plt.show()
```
